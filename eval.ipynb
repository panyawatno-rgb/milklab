```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from app import retrieve_top_k, chunks

# 1. Ground Truth 10 คำถาม
ground_truth = [
    {"query": "ร้านเปิดกี่โมง", "relevant_chunk_ids": [0]},
    {"query": "นมเหนียวดาร์กช็อกโกแลตราคาเท่าไหร่", "relevant_chunk_ids": [0]},
    {"query": "มีเมนูอะไรแนะนำบ้าง", "relevant_chunk_ids": [0]},
    {"query": "รสชาติหวานมากไหม", "relevant_chunk_ids": [0]},
    {"query": "ร้านปิดวันไหน", "relevant_chunk_ids": [0]},
    {"query": "ขอเมนูช็อกโกแลตหน่อย", "relevant_chunk_ids": [0]},
    {"query": "ติดต่อร้านได้อย่างไร", "relevant_chunk_ids": [0]},
    {"query": "มีส่วนผสมของนมไหม", "relevant_chunk_ids": [0]},
    {"query": "ราคาเริ่มต้นกี่บาท", "relevant_chunk_ids": [0]},
    {"query": "นมเหนียวคืออะไร", "relevant_chunk_ids": [0]},
]

precisions, recalls = [], []

for item in ground_truth:
    q = item["query"]
    gt_ids = set(item["relevant_chunk_ids"])
    
    retrieved = retrieve_top_k(q, k=3)
    retrieved_ids = set([chunks.index(c) for c in retrieved if c in chunks])
    
    hits = len(retrieved_ids.intersection(gt_ids))
    
    p3 = hits / 3.0
    r3 = hits / len(gt_ids) if len(gt_ids) > 0 else 0
    
    precisions.append(p3)
    recalls.append(r3)

print(f"Mean Precision@3: {np.mean(precisions):.4f}")
print(f"Mean Recall@3: {np.mean(recalls):.4f}")

# 2. Plot Histogram
plt.figure(figsize=(6, 4))
plt.hist(precisions, bins=5, color='skyblue', edgecolor='black')
plt.title("Precision@3 Distribution")
plt.xlabel("Precision@3")
plt.ylabel("Count")
plt.show()
```
